```python
import ast
import pandas as pd

def get_names(q):
    try:
        return [d['type']['displayName'] for d in ast.literal_eval(q)]
    except Exception:
        return []

def zone_profile(name):
    t = df_events[(df_events['player'] == name) & (df_events['is_touch'])].copy()
    t['q_names'] = t['qualifiers'].apply(get_names)
    t = t[~t['q_names'].apply(lambda n: 'CornerTaken' in n)]
    thirds = pd.cut(t['x'], [0, 100/3, 200/3, 100],
                    labels=['Own', 'Middle', 'Final']).value_counts(normalize=True).round(2)
    return {'n': len(t), **thirds.to_dict(),
            'right lane': round((t['y'] < 100/3).mean(), 2)}
```

```python
for p in ['Lamine Yamal', 'Pedro Porro']:
    print(p, zone_profile(p))

for p in ['Lamine Yamal', 'Pedro Porro']:
    t = df_events[(df_events['player'] == p) & (df_events['is_touch'])]
    print(p, 'posisi rata-rata x:', round(t['x'].mean(), 1), 'y:', round(t['y'].mean(), 1))
```
